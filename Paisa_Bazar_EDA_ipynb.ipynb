### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
#Removing all the rows, where Num_of_loan is 0
data_pb = data_pb[data_pb['Num_of_Loan'] != 0]
data_pb
````

````python
month_mapping = {
    1: "January", 2: "February", 3: "March", 4: "April",
    5: "May", 6: "June", 7: "July", 8: "August"
}

# Replace numerical values with month names
data_pb['Month'] = data_pb['Month'].map(month_mapping)
````

````python
data_pb
````

````python
# Removing all the rows, where Payment_of_Min_Amount is NM
data_pb = data_pb[data_pb['Payment_of_Min_Amount'] != 'NM']
````

````python
# Removing all the rows, where num_of_loanis 1 and type_of_loan is Not specified
data_pb = data_pb.loc[~((data_pb['Num_of_Loan'] == 1) & (data_pb['Type_of_Loan'] == 'Not Specified'))]
````

````python
# Define the function to identify outliers
def outliers(df, column_name):
  # Calculate Q1 using the correct column name
  Q1 = df[column_name].quantile(0.25)
  # Calculate Q3 using the correct column name
  Q3 = df[column_name].quantile(0.75)
  # Calculate the Interquartile Range (IQR)
  IQR = Q3 - Q1
  # Calculate the lower limit for outlier detection
  lower_limit = Q1 - 1.5 * IQR
  # Calculate the upper limit for outlier detection
  upper_limit = Q3 + 1.5 * IQR
  # Filter the DataFrame to exclude outliers in the specified column
  # Corrected the column access for filtering
  data = df[(df[column_name] > lower_limit) & (df[column_name] < upper_limit)]
  # Return the DataFrame without outliers
  return data

# Call the function with the DataFrame and the column name as a string
for i in ['Annual_Income',
       'Monthly_Inhand_Salary','Outstanding_Debt', 'Credit_Utilization_Ratio',
       'Total_EMI_per_month', 'Amount_invested_monthly','Monthly_Balance']:
       outliers(data_pb, i)
````

````python
for i, predictor in enumerate(data_pb[['Annual_Income',
       'Monthly_Inhand_Salary','Outstanding_Debt', 'Credit_Utilization_Ratio',
       'Total_EMI_per_month', 'Amount_invested_monthly','Monthly_Balance']]):
    plt.figure(i)
    sns.boxplot(data=data_pb,y=predictor)
````
